```python
from pathlib import Path
from src.utils import read_yaml, create_directories
from src.constant import CONFIG_FILE_PATH, PARAMS_FILE_PATH
from src.entity import DataingestionConfig  

class ConfigurationManager:
    def __init__(self,
                 config_filepath=CONFIG_FILE_PATH,
                 params_filepath=PARAMS_FILE_PATH):
        
        self.config = read_yaml(Path(config_filepath))
        self.params = read_yaml(Path(params_filepath))

        self.artifacts_root = Path(self.config.artifacts_root)
        create_directories([self.artifacts_root])

    def get_dataingestion_config(self) -> DataingestionConfig:
        ingestion_config = self.config.dataingestion

        ingestion_root = Path(ingestion_config.root_dir)
        create_directories([ingestion_root])

        return DataingestionConfig(
            root_dir=ingestion_root,
            source_URL=ingestion_config.source_URL,
            local_data_file=Path(ingestion_config.local_data_file),
            unzip_dir=Path(ingestion_config.unzip_dir)
        )
```

```python
import os
import zipfile
import urllib.request

class DataIngestion:
    def __init__(self, config):
        self.config = config

    def download_file(self, url, local_path):
        try:
            # Download file if it doesn't exist
            if not os.path.exists(local_path):
                print(f"Downloading file from {url} to {local_path}...")
                urllib.request.urlretrieve(url, local_path)
                print("Download completed.")
            else:
                print(f"File already exists: {local_path}")
        except Exception as e:
            print(f"Error downloading file: {str(e)}")
            raise e

    def extract_zip_file(self):
        unzip_path = self.config.unzip_dir
        # Ensure the unzip directory exists
        os.makedirs(unzip_path, exist_ok=True)

        # Check if the local data file exists
        if not os.path.exists(self.config.local_data_file):
            print(f"File {self.config.local_data_file} not found. Downloading...")
            self.download_file(self.config.source_URL, self.config.local_data_file)

        # Extract zip file
        try:
            print(f"Extracting {self.config.local_data_file} to {unzip_path}")
            with zipfile.ZipFile(self.config.local_data_file, 'r') as zip_ref:
                zip_ref.extractall(unzip_path)
                print(f"Extraction complete at {unzip_path}")
        except Exception as e:
            print(f"Error extracting zip file: {str(e)}")
            raise e
```
